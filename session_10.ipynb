```python
# 1. Install Matplotlib in your Python environment and create a simple line plot showing the number 
# of daily steps you took over the last 7 days. Save the chart as steps_lineplot.png.
import matplotlib
print(matplotlib.__version__)

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [5000, 6500, 7000, 4500, 8000, 9000, 7500]

plt.plot(days,steps)
plt.savefig("steps_lineplot.png")
```

```python
# 2. Using Matplotlib, create a scatter plot of 10 restaurants from your city with their Zomato ratings
#  on the x-axis and average meal price on the y-axis. Add axis labels and a title to the chart.


ratings = [4.5, 4.2, 3.8, 4.7, 4.0, 3.5, 4.6, 4.1, 3.9, 4.4]

prices = [800, 600, 400, 1200, 500, 300, 1000, 550, 450, 750]

plt.scatter(ratings,prices)
plt.title("Ahemdabad city restaurants")
plt.xlabel("zometo rating")
plt.ylabel("average of meal price")

```

```python
# 3.# Build a bar chart that shows the number of orders you or your friends placed on Swiggy, Zomato, 
# and Domino’s in the last month. Use different colors for each bar and add a legend.

apps = ['zomato','swiggy',"domino's"]

order = [30,20,33]

plt.figure(figsize=(6,6))

plt.bar(apps[0], order[0], color='red', label='Zomato')
plt.bar(apps[1], order[1], color='yellow', label='Swiggy')
plt.bar(apps[2], order[2], color='green', label='Dominos')
plt.xlabel("apps")
plt.ylabel("order")
plt.legend(['zomato','swiggy','Dominos'])
```

```python
# 4.Create a histogram of the durations (in minutes) of your last 20 Spotify listening sessions using Matplotlib. 
# Set the number of bins to 5 and customize the color of the bars.

duration = [12, 25, 30, 18, 45, 50, 22, 35, 40, 15,
            28, 32, 55, 20, 42, 38, 27, 48, 10, 33]


plt.hist(duration,bins=5,color='orange')
plt.xlabel("Listening Duration (minutes)")
plt.ylabel("Number of Sessions")

plt.title("Spotify Listening Session Durations")
```
