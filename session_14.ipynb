```python
# 2.# Using a dataset of Flipkart product reviews (at least columns: 'rating', 'category'), create a bar plot showing
#  the count of reviews for each rating (1-5 stars) using matplotlib or seaborn.


df = pd.read_csv(r"D:\Tops Technowlogy\Data Scince\Dummy_Library_Assignment\csv_files\flipkart_reviews.csv")
sns.countplot(data = df , x= "rating")

rating_counts = df['rating'].value_counts().sort_index()

plt.bar(rating_counts.index, rating_counts.values)
```
