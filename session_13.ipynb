```python
# 2. Using the 'flights' dataset from Seaborn, generate a heatmap showing the correlation between months and years 
# based on the number of passengers.<br><br><em><strong>Hint:</strong> Use pivot_table to reshape the data before 
# plotting the heatmap.</em>

df = sns.load_dataset("flights")

pivots = df.pivot_table(
    index = "year",
    columns = "month",
    values = "passengers"
)
sns.heatmap(pivots)
```

```python
# 3.# Create a relplot using the 'fmri' dataset from Seaborn to visualize how the signal changes over time for 
# different event types.

fmri = sns.load_dataset("fmri")

sns.relplot(
    data = fmri,
    x="timepoint",
    y="signal",
    hue="event"
)
```

```python
# 4.Pick any categorical column from the 'titanic' dataset (like 'class' or 'sex') and use catplot to display the survival
#  rate for each group with confidence intervals.

titanic = sns.load_dataset("titanic")
sns.catplot(x="sex",y="survived",kind="bar",errorbar=("ci",95),data = titanic)
```

```python
# 5. Use jointplot on the 'penguins' dataset to visualize the relationship between bill_length_mm and flipper_length_mm,
#  and fit a regression line on the scatter plot.<br><br><em><strong>Hint:</strong> Set kind='reg' in jointplot for
#  regression line.</em>

penguins = sns.load_dataset("penguins")

sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="flipper_length_mm",
    kind="reg"
)
```
