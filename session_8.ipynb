```python
# 4. Suppose you have a DataFrame of Instagram usernames where some entries have typos (like 'insta_queen', 
#'insta-queen', 'instaqueen'). Use the replace() function to standardize all these variants to 'instaqueen'.


instagram = pd.read_csv(r'D:\Tops Technowlogy\Data Scince\Dummy_Library_Assignment\csv_files\instagram_usernames.csv')
instagram['username']=instagram['username'].replace({"insta_queen":'instaqueen',
                                                     'insta-queen':'instaqueen',
                                                     'instaqueen':'instaqueen',
                                                     'Insta.Queen': 'instaqueen',
                                                     'INSTAQUEEN': 'instaqueen'})

instagram.head(30)
```

```python
# 5. You have a DataFrame column for payment status from a Paytm-like app with mixed values: 
# 'Yes', 'yes', 'Y', 'No', 'no', 'N', and some with extra spaces. Write code to unify this column 
# so all paid statuses become 1 and all unpaid statuses become 0, trimming whitespace and fixing 
# capitalization where needed.<br><br><em><strong>Hint:</strong> Use str.strip(), str.lower(), 
# and map/replace methods in pandas.</em>



payment = pd.read_csv(r"D:\Tops Technowlogy\Data Scince\Dummy_Library_Assignment\csv_files\paytm_transactions.csv")
payment['payment_status'] =payment['payment_status'].str.strip().str.lower().map({'yes':1,'y':1,'no':0,'n':0})
payment['payment_status']
```
