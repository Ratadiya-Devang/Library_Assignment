```python
# 2. Load a CSV file containing order dates from a Flipkart-style order history (column: 'order_date', 
# format: 'YYYY-MM-DD HH:MM:SS'). Extract the year, month, and weekday for each order and add them as 
# new columns in the DataFrame.

flipkart = pd.read_csv(r'D:\Tops Technowlogy\Data Scince\Dummy_Library_Assignment\csv_files\flipkart_order_history.csv')
flipkart['order_date'] =pd.to_datetime(flipkart['order_date'])
flipkart['year'] = flipkart['order_date'].dt.year
flipkart['month'] = flipkart['order_date'].dt.month
flipkart['day'] = flipkart['order_date'].dt.day
flipkart['week day']=flipkart['order_date'].dt.day_of_week

flipkart
```

```python
# 3. Set the 'order_date' column as the index of your DataFrame and use resampling to calculate the total 
# number of orders placed each week.<br><br><em><strong>Hint:</strong> Use df.resample('W').size() after 
# setting the datetime index.</em>


flipkart_timeseries = pd.read_csv(r'D:\Tops Technowlogy\Data Scince\Dummy_Library_Assignment\csv_files\flipkart_orders_timeseries.csv')

flipkart_timeseries['order_date']=pd.to_datetime(flipkart_timeseries['order_date'])
flipkart_timeseries.set_index("order_date",inplace=True)
week_order = flipkart_timeseries.resample("W").size()
week_order
```

```python
# 4. Suppose you have a DataFrame of Instagram posts with a 'posted_at' column in UTC.
#  Convert these timestamps to 'Asia/Kolkata' timezone and display the first 5 converted times.

timezone = pd.read_csv(r'D:\Tops Technowlogy\Data Scince\Dummy_Library_Assignment\csv_files\instagram_posts.csv')
timezone['posted_at'] = pd.to_datetime(timezone['posted_at'])
timezone['indian_time'] = timezone['posted_at'].dt.tz_convert("Asia/Kolkata")
timezone
```
