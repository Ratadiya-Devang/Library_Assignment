```python
# 2.Simulate a Spotify-like 'Recommended Songs' feature: Given two 3x3 matrices representing user-song 
# interaction scores (user preferences and song popularity), use dot() and matmul() to compute the final 
# recommendation matrix and explain the difference between the two results.


import numpy as np

user_preferences = np.array([
    [5, 2, 1],
    [1, 5, 2],
    [2, 1, 5]
])


song_popularity = np.array([
    [4, 1, 2],
    [2, 5, 1],
    [1, 2, 4]
])

recommendation_dot = np.dot(user_preferences, song_popularity)

recommendation_matmul = np.matmul(user_preferences, song_popularity)

print("Recommendation using dot():")
print(recommendation_dot)

print("\nRecommendation using matmul():")
print(recommendation_matmul)
```

```python
# 3.Given a 4x4 NumPy matrix representing the pixel brightness of a small Instagram image, 
# use transpose (T) to rotate the image and then calculate the mean, median, standard deviation,
# and variance of the pixel values.

a = np.random.randint(300,1000,(4,4))
print(a)
print(np.mean(a))
print(np.median(a))
print(np.std(a))
print(np.var(a))

print("\n")

Transpose = a.transpose()
print(Transpose)
print(np.mean(Transpose))
print(np.median(Transpose))
print(np.std(Transpose))
print(np.var(Transpose))
```

```python
# Take a 3x3 NumPy matrix representing a Zomato restaurant rating correlation grid and use np.linalg.inv(),
#  np.linalg.det(), and np.linalg.eig() to compute its inverse, determinant, and eigenvalues/eigenvectors.
#  <br><br><em><strong>Hint:</strong> If the matrix is not invertible, modify one value and try again.</em>


import numpy as np

A = np.array([
    [5, 2, 1],
    [2, 4, 2],
    [1, 2, 5]
])

det = np.linalg.det(A)

inverse = np.linalg.inv(A)

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix:")
print(A)

print("Determinant:")
print(det)

print("Inverse:")
print(inverse)

print("Eigenvalues:")
print(eigenvalues)

print("Eigenvectors:")
print(eigenvectors)
```
