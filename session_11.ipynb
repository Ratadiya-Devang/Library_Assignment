```python
# 2. Plot a comparison of average delivery times for Zomato, Swiggy, and Domino's using a bar chart, and style 
# each bar with a different color, linestyle, and linewidth using Matplotlib plot styling options.

apps = ["Zomato", "Swiggy", "Dominos"]
delivery_time = [35, 30, 25]

bars = plt.bar(apps,delivery_time,color=['red','green','blue'], edgecolor=["black", "black", "black"],linewidth=[2, 3, 4])
bars[0].set_linestyle("--")
bars[1].set_linestyle("-")
bars[2].set_linestyle(":")
```

```python
# 3. Create a multi-axis chart that shows the number of Instagram followers (left y-axis) and average daily posts 
# (right y-axis) for five influencers. Use different colors and markers for each axis.
import matplotlib.pyplot as plt

influencers = ["A", "B", "C", "D", "E"]

followers = [50000, 80000, 120000, 200000, 150000]

daily_posts = [3, 5, 2, 7, 4]

fig,ax1 = plt.subplots()
ax1.plot(influencers,followers,color="red")
ax1.set_xlabel("influencers")
ax1.set_ylabel("followers")

ax2 = ax1.twinx()
ax2.plot(influencers,daily_posts)
ax2.set_ylabel("posts")

```

```python
# 4. Plot the number of tickets sold for five recent Bollywood movies (categorical) versus their IMDB ratings 
# (numeric) using a scatter plot. Add annotations to display the movie names above each point.

movies = ["Movie A", "Movie B", "Movie C", "Movie D", "Movie E"]

tickets = [50000, 75000, 40000, 90000, 60000]

ratings = [7.2, 8.1, 6.8, 8.5, 7.5]

plt.scatter(tickets, ratings, color="blue")

for i in range(len(movies)):
    plt.annotate(
        movies[i],
        (tickets[i],ratings[i]),
        xytext = (0,10),
         textcoords="offset points"
    )

```

```python
# 5.# Add a custom text annotation to a Matplotlib chart showing Flipkart's monthly sales, marking the highest sales 
# point with the label 'Big Billion Days'.<br><br><em><strong>Hint:</strong> Use the ax.annotate() function to place 
# the label at the correct data point.</em>

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 120, 90, 150, 300, 180]

fig, ax = plt.subplots()

ax.plot(months, sales, marker="o")

max_sales = max(sales)
max_index = sales.index(max_sales)

ax.annotate(
    "Big Billion Days",
    (months[max_index], sales[max_index]),
    xytext=(0, 20),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

ax.set_xlabel("Month")
ax.set_ylabel("Monthly Sales")
ax.set_title("Flipkart Monthly Sales")

plt.show()
```
