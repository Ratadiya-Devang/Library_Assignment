```python
# 5. Use ChatGPT or Copilot to generate a Python code snippet that reads a semicolon-separated CSV 
# of Paytm transactions, detects missing values, and prints out which columns have nulls. Test the 
# code with a small sample file and fix any errors you encounter.

data = pd.read_csv('paytm_transactions_sample.csv',delimiter=';')
missing_value_sum = data.isnull().sum()

for i in missing_value_sum.index:
    if(missing_value_sum[i]>0):
        print(i,"-->",missing_value_sum[i],'missing')
```
