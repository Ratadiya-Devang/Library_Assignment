```python
import pandas as pd 
import numpy as np
from sqlalchemy import create_engine
```

```python
# 1.# Download a sample CSV file of IPL player stats (include columns like player name, runs, matches, 
# and some missing values). Load it into a pandas DataFrame and use isnull() and notnull() to print
#  how many missing values are present in each column.


df = pd.read_csv("C:/Users/LENOVO/Downloads/ipl_player_stats.csv")
null = df.isnull().sum()
notnull = df.notnull().sum()

print("null value in data is :== \n",null)

print("\n")

print("not null value in data is :== \n",notnull)
```

```python
# 2. Using the loaded IPL player stats DataFrame, apply dropna(axis=0, how='any') to remove all 
# rows with any missing data and display the shape of the DataFrame before and after dropping.

print("before droping null value data shape is :== ",df.shape)

df.dropna(axis=0,how='any',inplace = True)
print("after droping null value data shape is :== ",df.shape)
```

```python
# 3.# For the 'runs' column in your IPL player stats DataFrame, use fillna() to replace missing values
#  with the mean of the column. Print the updated column to verify the changes.

print("before runs empty data = ",df["runs"].isna().sum())
df['runs'].fillna(df['runs'].mean(), inplace=True)
print("after runs is fill = ",df["runs"].isna().sum())
```
